# PCA on Breast Cancer Gene Expression Data

This project performs Principal Component Analysis (PCA) on breast cancer gene-expression data.

Genes analyzed:
- GATA3
- XBP1

The PCA projection demonstrates separation between ER+ and ER− breast cancer samples.

The analysis reproduces the biological interpretation of Figure 1 from the referenced breast cancer study.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
# Load class labels
class_df = pd.read_csv(
    "../data/class.tsv",
    sep="\t",
    header=None
)

# Load gene expression data
expression_df = pd.read_csv(
    "../data/filtered.tsv.gz",
    sep="\t",
    compression="gzip"
)

# Load gene mapping file
columns_df = pd.read_csv(
    "../data/columns.tsv.gz",
    sep="\t",
    compression="gzip",
    comment="#"
)
```

```python
import gzip

with gzip.open(
    "../data/columns.tsv.gz",
    "rt"
) as file:

    for i in range(10):
        print(file.readline())

print(columns_df.head())
print(columns_df.columns)
columns_df[
    columns_df["GeneSymbol"].isin(
        ["XBP1", "GATA3"]
    )
]
expression_df.columns = (
    expression_df.columns.str.strip()
)
gata3 = expression_df["4359"]

xbp1 = expression_df["4404"]

labels = class_df[0]
```

```python
fig, ax = plt.subplots(figsize=(8,6))

for label in [0, 1]:

    mask = labels == label

    ax.scatter(
        gata3[mask],
        xbp1[mask],
        label=f"Class {label}"
    )

ax.set_xlabel("GATA3 Expression")

ax.set_ylabel("XBP1 Expression")

ax.set_title("Breast Cancer Gene Expression")

ax.legend()

ax.grid(True)

fig.savefig(
    "../figures/figure1a.png",
    dpi=300
)

plt.show()
```

```python
selected_data = pd.DataFrame({
    "GATA3": gata3,
    "XBP1": xbp1
})
scaler = StandardScaler()

scaled_data = scaler.fit_transform(
    selected_data
)
pca = PCA(n_components=1)

pc1 = pca.fit_transform(
    scaled_data
)
fig, ax = plt.subplots(figsize=(10,4))

for label in [0, 1]:

    mask = labels == label

    ax.scatter(
        pc1[mask],
        np.zeros(sum(mask)),
        label=f"Class {label}"
    )

ax.set_xlabel("Principal Component 1")

ax.set_title("Projection on PC1")

ax.legend()

ax.grid(True)

fig.savefig(
    "../figures/figure1c.png",
    dpi=300
)

plt.show()
```

```python
print(
    "Explained Variance Ratio:"
)

print(
    pca.explained_variance_ratio_
)

```
